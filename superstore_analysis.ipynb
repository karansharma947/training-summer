## Day 3: Data Cleaning


```python
# Create a copy for cleaning
superstore_clean = data.copy()

# Store values before cleaning
missing_before = superstore_clean.isnull().sum()
duplicates_before = superstore_clean.duplicated().sum()
rows_before = superstore_clean.shape[0]

print("Rows before cleaning:", rows_before)
print("Missing values before cleaning:")
print(missing_before[missing_before > 0])
print("Duplicate rows before cleaning:", duplicates_before)
```

```python
# Detecting Outliers
numeric_cols=data.select_dtypes(include="number").drop(columns=["Sales","Quantity"]).columns
for column in numeric_cols:
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    
    IQR = Q3 - Q1
    
    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR
    
    outliers = data[(data[column] < lower_limit) | (data[column] > upper_limit)]
    
    print("Column:", column)
    print("Lower Limit:", lower_limit)
    print("Upper Limit:", upper_limit)
    print("Number of Outliers:", outliers.shape[0])
    print("Min value:",data[column].min())
    print("Max value:",data[column].max())

    print("-" * 40)

# we are dealing with quantity and sales separately because these values cannot be negative

cols=["Quantity","Sales"]
for column in cols:
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    
    IQR = Q3 - Q1
    
    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    lower_limit=max(0,lower_limit)
    outliers = data[(data[column] < lower_limit) | (data[column] > upper_limit)]
    
    print("Column:", column)
    print("Lower Limit:", lower_limit)
    print("Upper Limit:", upper_limit)
    print("Number of Outliers:", outliers.shape[0])
    print("Min value:",data[column].min())
    print("Max value:",data[column].max())

    print("-" * 40)
```

```python
#  Handle missing values

# Fill missing Customer Name with Unknown
superstore_clean["Customer Name"] = superstore_clean["Customer Name"].fillna("Unknown")

superstore_clean = superstore_clean.dropna(subset=["Postal Code"])

# Fill missing numerical values using median
for col in ["Sales", "Profit"]:

    superstore_clean[col] = superstore_clean[col].fillna(superstore_clean[col].median())

print("Missing values after treatment:")
print(superstore_clean.isnull().sum())
```

```python
# 2. Remove duplicate records
superstore_clean = superstore_clean.drop_duplicates()
rows_after_duplicates = superstore_clean.shape[0]

print("Duplicate rows found before cleaning:", duplicates_before)
print("Rows remaining after duplicate removal:", rows_after_duplicates)
print("Duplicate rows after cleaning:", superstore_clean.duplicated().sum())
```

```python
# 3. Check and correct data types
print("Data types before date conversion:")
print(superstore_clean[["Order Date", "Ship Date"]].dtypes)
print()

# Convert date columns into datetime
superstore_clean["Order Date"] = pd.to_datetime(superstore_clean["Order Date"], format="mixed")
superstore_clean["Ship Date"] = pd.to_datetime(superstore_clean["Ship Date"], format="mixed")


print("Data types after conversion:")
print(superstore_clean[["Order Date", "Ship Date"]].dtypes)
```

```python
# 4. Clean text columns by removing extra spaces
text_cols = superstore_clean.select_dtypes(include=["object", "string"]).columns
for col in text_cols:
    superstore_clean[col] = superstore_clean[col].str.strip().str.title()
print("Text cleaning completed for these columns:")
print(list(text_cols))
```

```python
# 5. Verify the cleaned dataset
print("Cleaned Dataset Information:")
superstore_clean.info()
print()

print("Missing values after cleaning:")
print(superstore_clean.isnull().sum())
print()

print("Duplicate rows after cleaning:", superstore_clean.duplicated().sum())
```

```python
# Data Cleaning Report
rows_after_cleaning = superstore_clean.shape[0]

print("DATA CLEANING REPORT")
print("=" * 50)

print("1. Missing Value Treatment")
print("Missing columns before cleaning:")
print(missing_before[missing_before > 0])
print("Missing values were handled using fillna() and dropna().")
print()

print("2. Duplicate Removal Summary")
print("Duplicate rows found:", duplicates_before)
print("Rows before cleaning:", rows_before)
print("Rows after cleaning:", rows_after_cleaning)
print()

print("3. Data Type Corrections")
print("Order Date and Ship Date were converted to datetime.")
print("Postal Code was converted to integer after removing missing values.")
print()

print("4. Text Cleaning Summary")
print("Extra spaces were removed from text columns using str.strip().")
print()

print("5. Final Conclusion")
print("1. Missing values have been handled.")
print("2. Duplicate rows have been removed.")
print("3. Date columns are now in datetime format.")
print("4. Text columns are cleaned.")
print("5. The dataset is ready for feature engineering and analysis.")
print("6. The data loss cleaning.",(len(data)-len(superstore_clean))/len(data)*100)

```

## Day 4: Feature Engineering


```python

# 2. Create Order Year
superstore_clean["Order Year"] = superstore_clean["Order Date"].dt.year

# 3. Create Order Month
superstore_clean["Order Month"] = superstore_clean["Order Date"].dt.month_name()
superstore_clean["Order Month Number"] = superstore_clean["Order Date"].dt.month

# 4. Create Delivery Days
superstore_clean["Delivery Days"] = (superstore_clean["Ship Date"] - superstore_clean["Order Date"]).dt.days

# 5. Create Profit Margin
superstore_clean["Profit Margin"] = (superstore_clean["Profit"] / superstore_clean["Sales"]) * 100

# 6. Create Profit Status
superstore_clean["Profit Status"] = np.where(superstore_clean["Profit"] > 0, "Profit", "Loss")

# Display first 5 rows with new columns
superstore_clean.head()
```

```python
# Day 4 Questions and Answers
most_orders_year = superstore_clean["Order Year"].value_counts().idxmax()
most_orders_month = superstore_clean["Order Month"].value_counts().idxmax()
avg_delivery_time = superstore_clean["Delivery Days"].mean()
avg_profit_margin = superstore_clean["Profit Margin"].mean()
loss_making_orders = superstore_clean[superstore_clean["Profit Status"] == "Loss"]

print("1. Year with most orders:", most_orders_year)
print("2. Month with most orders:", most_orders_month)
print("3. Average delivery time:", avg_delivery_time)
print("4. Number of loss-making orders:", loss_making_orders.shape[0])
print("5. Average profit margin:", avg_profit_margin)
print()

print("Loss-making orders sample:")
loss_making_orders[["Order ID", "Category", "Sub-Category", "Sales", "Profit", "Profit Status"]].head()
```

```python
# Day 4 Observations
print("DAY 4 OBSERVATIONS")
print("=" * 50)
print("1. The year with the most orders is", most_orders_year)
print("2. The month with the most orders is", most_orders_month)
print("3. The average delivery time is", round(avg_delivery_time, 2), "days.")
print("4. There are", loss_making_orders.shape[0], "loss-making orders.")
print("5. The average profit margin is", round(avg_profit_margin, 2), "%.")
```
